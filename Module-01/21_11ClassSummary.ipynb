# Python Class Summary Notes: Conditionals & Functions

## 1. Numeric Input (`float()`) & F-String Formatting
* **`float()` Conversion:** `input()` always returns a string (`str`). Wrapping it in `float()` converts text into decimal numbers, which is essential for money, percentages, and math operations.
* **String Concatenation Pitfall:** Adding unconverted strings joins them together instead of doing math (e.g., `"2" + "10"` produces `"210"`, not `12`).
* **Decimal Formatting (`:.2f`):** Inside an f-string, `:.2f` formats a number to display exactly 2 decimal places (e.g., `${total_compensation:.2f}`).

```python
salary = float(input("What is your salary? "))
print(f"Total compensation is: ${salary:.2f}")
```

---

## 2. Comparison & Logical Operators
* **Comparison Operators:** Evaluate two values and return a boolean (`True` or `False`):
  * `==` (equal to) and `!=` (not equal to)
  * `>` (greater than) and `<` (less than)
  * `>=` (greater than or equal to) and `<=` (less than or equal to)
* **Logical Operators:** Combine multiple conditions:
  * **`and`:** Both conditions must evaluate to `True`.
  * **`or`:** At least one condition must evaluate to `True`.

---

## 3. Conditional Statements (`if`, `elif`, `else`)
* **Purpose:** Execute different blocks of code based on whether conditions evaluate to `True` or `False`.
* **Top-Down Evaluation:** Python checks conditions from top to bottom and runs **only the first branch** that evaluates to `True`, skipping the rest. (This is why grading checks `>= 70` before `>= 60`).

```python
if score >= 70:
    print("Grade A")
elif score >= 60:
    print("Grade B")
elif score >= 50:
    print("Grade C")
elif score >= 40:
    print("Grade D")
else:
    print("Grade F")
```

---

## 4. Defining Functions: `print()` vs. `return`
* **`def` Keyword:** Defines a reusable block of code that takes parameters (inputs) inside parentheses.
* **Why `return` Matters:**
  * A function that only uses `print()` displays text on the screen, but hands back `None` (`NoneType`) to the program. Trying to do math with `None` causes a `TypeError`.
  * Using `return` sends the actual calculated value back to the caller so it can be stored in a variable or reused in further calculations.

```python
def add_numbers(num1, num2):
    return num1 + num2

result = add_numbers(5, 3)  # Stores 8 for reuse
```

---

## 5. Combining Functions & Conditionals (Boolean Return)
* **Wrapping Logic:** Conditional checks can live inside functions to evaluate inputs cleanly.
* **Direct Boolean Return:** Because comparison expressions like `number % 2 == 0` already evaluate to `True` or `False`, you can return the expression directly without needing an `if/else` block.

```python
def is_even(number):
    return number % 2 == 0  # Returns True if remainder is 0, else False
```

## Part 1: Compensation Calculator (`if/else`, `float()`, & `and`)
* **How it works:** `input()` captures user entries as text, so `float()` converts them into decimals for money and percentages.
* **Logic check:** The `and` operator ensures **both** `bonus` and `commission_rate` stay within their thresholds before calculating and printing total compensation formatted to 2 decimal places (`:.2f`).

````python
print("--- Part 1: Compensation Calculator ---")
bonus_threshold = 30000
commission_rate_threshold = 0.3
total_comp_threshold = 200000

name = input("What is your name? ").strip()
salary = float(input("What is your salary? "))
bonus = float(input("What is your bonus? "))
commission_rate = float(input("What is your commission rate (as a decimal)? "))

total_compensation = salary + bonus + (salary * commission_rate)

if (
    bonus <= bonus_threshold
    and commission_rate <= commission_rate_threshold
    and total_compensation <= total_comp_threshold
):
    print(f"Total compensation for {name or 'Employee'} is: ${total_compensation:.2f}")
else:
    print(f"Salary entry for {name or 'Employee'} exceeds the threshold.")
````

## Part 2: Grade Evaluator (`if`, `elif`, `else`)
* **How it works:** Python evaluates conditions from top to bottom and stops at the **first** condition that evaluates to `True`.
* **Fallback (`else`):** Adding an `else` block catches any score below `40` so the program always gives feedback.


````python
score = 65

if score >= 70:
    print("Grade A")
elif score >= 60:
    print("Grade B")
elif score >= 50:
    print("Grade C")
elif score >= 40:
    print("Grade D")
else:
    print("Grade F")
````

## Part 3: Functions (`print()` vs. `return`)
* **Why `return` matters:** A function that only uses `print()` shows output on screen, but hands back `None` (`NoneType`), which cannot be reused in math.
* **Numeric conversion:** Always convert `input()` with `float()` (or `int()`) before adding, otherwise `"2" + "10"` joins as `"210"`.

````python
def add_numbers(num1, num2):
    return num1 + num2

# Reusing returned values in further math
first_sum = add_numbers(5, 3)
second_sum = add_numbers(10, 7)
print("Combined total:", first_sum + second_sum)

# Converting user input before calling the function
x = float(input("Enter the first number: "))
y = float(input("Enter the second number: "))
print(f"The sum of {x} and {y} is: {add_numbers(x, y)}")
````

## Part 4: Wrapping Conditionals Inside Functions
* **How it works:** Comparison operators (`==`, `!=`, `>`, `<`, `>=`, `<=`) already evaluate to `True` or `False`.
* **Clean shortcut:** Instead of writing a 4-line `if/else` block inside `is_even()`, you can return the comparison `number % 2 == 0` directly.


````python
def is_even(number):
    return number % 2 == 0

print("Is 4 even?", is_even(4))  # True
print("Is 5 even?", is_even(5))  # False
````
