## 1. `while` Loops & Infinite Loop Prevention
* **How it works:** Repeats a block of code as long as its condition evaluates to `True`.
* **Avoiding Infinite Loops:** You must update the loop variable inside the loop (e.g., `number -= 1` or `week += 1`) so the condition eventually becomes `False`.
* **Formatting Numbers (`:,.2f`):** Inside an f-string, `:,` adds thousand-separator commas while `.2f` fixes 2 decimal places (e.g., `$25,000.00`).

```python
# Countdown example
sound = "Bark"
number = 3

while number > 0:
    print(sound)
    number -= 1  # Decrement to avoid an infinite loop
    print(f"Remaining count: {number}")

print("-" * 30)

# Savings target accumulator using main() and helper function
def weeks_to_target(target, weekly):
    if weekly <= 0:
        return 0  # Guard against infinite loop if weekly savings isn't positive
    saved = 0
    week = 0
    while saved < target:
        week += 1
        saved += weekly
        print(f"Week {week}: ${saved:,.2f}")
    return week

def main():
    target = 200000
    weekly = 25000
    weeks = weeks_to_target(target, weekly)
    print(f"Adesola hits his target in {weeks} weeks.")

main()
```

## 2. `for` Loops & Iterables
* **How it works:** Iterates over each item in an iterable sequence (such as a `list`, `str`, `tuple`, `dict`, or `set`) and runs the indented code block once per item.
* **Strings as Iterables:** Looping through `"ADESOLA"` runs 7 times—once for each character in the string.

```python
# Looping over a list (5 iterations)
numbers = [1, 2, 3, 4, 5]
for num in numbers:
    print("BOLE IS AVAILABLE!")

print("-" * 30)

# Looping over a string (7 characters = 7 iterations)
for char in "ADESOLA":
    print(f"Character: {char} -> BOLE IS AVAILABLE!")
```

## 3. Lists & Mutability
* **Definition:** Created using square brackets `[]` (e.g., `states = ["Lagos", "Ogun", "Oyo", "Ondo", "Ekiti"]`).
* **Mutable:** Lists can be modified in place after creation:
  * **Update by index:** `states[0] = "Abuja"`
  * **Add item:** `states.append("Osun")`
  * **Remove item:** `states.remove("Ogun")`

```python
states = ["Lagos", "Ogun", "Oyo", "Ondo", "Ekiti"]

for state in states:
    print(state.upper())

# Demonstrating list mutability (modifying, adding, removing)
states[0] = "Abuja"
states.append("Osun")
states.remove("Ogun")
print("Updated states list:", states)
```

## 4. Positive Indexing, Negative Indexing & Slicing
* **Zero-Based Indexing:** The first item is at index `0` (`states[0]` -> `"Lagos"`).
* **Negative Indexing:** Counts backward from the end of a sequence:
  * `-1` gets the last item (`"MARKSON"[-1]` -> `"N"`).
  * `-3` gets the third-to-last item (`states[-3]` -> `"Oyo"`).
* **Slicing (`[start:stop]`):** Extracts a portion of a sequence, **including `start` but excluding `stop`** (`"Lagos"[0:2]` -> `"La"`).

```python
name = "MARKSON"
fresh_states = ["Lagos", "Ogun", "Oyo", "Ondo", "Ekiti"]

print("Last letter of MARKSON:", name[-1])                 # 'N'
print("3rd-to-last state:", fresh_states[-3])              # 'Oyo'
print("First 2 letters of Lagos:", fresh_states[0][0:2])   # 'La'
print("First 3 states slice:", fresh_states[0:3])          # ['Lagos', 'Ogun', 'Oyo']
```
